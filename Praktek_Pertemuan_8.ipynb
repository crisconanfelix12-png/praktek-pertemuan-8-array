```python
# Ganti dengan list `energi` (30 elemen) dari blok kode di atas
energi = [
    120, 125, 118, 130, 128, 122, 126,
    115, 119, 121, 117, 124, 123, 120,
    132, 135, 129, 131, 127, 134, 130,
    110, 112, 108, 114, 116, 113, 115,
    125, 128
]

if len(energi) != 30:
    print(f"Peringatan: jumlah data saat ini {len(energi)} hari (diharapkan 30 hari).")

# 1. Total dan rata-rata energi bulanan
total_energi = sum(energi)
rata_rata_bulanan = total_energi / len(energi)

print("=== RINGKASAN BULANAN ===")
print(f"Total energi      : {total_energi:.2f}")
print(f"Rata-rata energi  : {rata_rata_bulanan:.2f}")

# 2. Tiga hari tertinggi dan tiga hari terendah (nomor hari 1-based dan nilainya)
data_bernomor = [(hari, nilai) for hari, nilai in enumerate(energi, start=1)]

tertinggi_3 = sorted(data_bernomor, key=lambda item: item[1], reverse=True)[:3]
terendah_3 = sorted(data_bernomor, key=lambda item: item[1])[:3]

print("\n=== 3 HARI TERTINGGI ===")
for hari, nilai in tertinggi_3:
    print(f"Hari ke-{hari}: {nilai}")

print("\n=== 3 HARI TERENDAH ===")
for hari, nilai in terendah_3:
    print(f"Hari ke-{hari}: {nilai}")

# 3. Jumlah hari di bawah target yang dimasukkan pengguna
target = float(input("\nMasukkan target energi harian: "))
hari_di_bawah_target = [hari for hari, nilai in data_bernomor if nilai < target]

print(f"Jumlah hari di bawah target ({target}): {len(hari_di_bawah_target)}")
if hari_di_bawah_target:
    print("Daftar hari:", hari_di_bawah_target)

# 4. Setiap hari yang rata-rata bergerak (moving average) 7 harinya < 97% rata-rata bulanan
ambang_ma7 = 0.97 * rata_rata_bulanan
print(f"\n=== MOVING AVERAGE 7 HARI < 97% RATA-RATA BULANAN ({ambang_ma7:.2f}) ===")

ditemukan_ma7 = False
for i in range(6, len(energi)):  # mulai indeks ke-6 = hari ke-7
    ma_7 = sum(energi[i - 6 : i + 1]) / 7
    if ma_7 < ambang_ma7:
        hari_ke = i + 1
        print(f"Hari ke-{hari_ke}: MA-7 = {ma_7:.2f}")
        ditemukan_ma7 = True

if not ditemukan_ma7:
    print("Tidak ada hari dengan rata-rata bergerak 7 hari di bawah ambang tersebut.")

# 5. Total energi per minggu (Minggu 1-4 masing-masing 7 hari, ditambah sisa hari)
print("\n=== TOTAL ENERGI PER MINGGU ===")
for minggu in range(4):
    awal = minggu * 7
    akhir = awal + 7
    total_minggu = sum(energi[awal:akhir])
    print(f"Minggu {minggu + 1} (Hari {awal + 1}-{akhir}): {total_minggu:.2f}")

if len(energi) > 28:
    sisa_energi = sum(energi[28:])
    print(f"Sisa hari (Hari 29-{len(energi)})   : {sisa_energi:.2f}")
```
